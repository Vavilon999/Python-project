````python
class Polynomial:

    def __init__(self,*c):

        if len(c) == 1 and isinstance(c[0], list):
            self.c = c[0].copy()

        elif len(c) == 1 and isinstance(c[0], dict):
            data = c[0]

            if not data:
                self.c = [0]
            else:
                maxd = max(data.keys())
                self.c = [0] * (maxd + 1)
            
            for d, ct in data.items():
                self.c[d] = ct

        elif len(c) == 1 and isinstance(c[0], Polynomial):
            self.c = c[0].c.copy()
        else: 
            self.c = list(c)
        if len(self.c) == 0:
            self.c = [0]

        while len(self.c) > 1 and self.c[-1] == 0:
            self.c.pop()

    def __repr__(self):
        return f"Polynomial {self.c}"

    def __str__(self):

        p = []
        
        for d in range(len(self.c) - 1, -1, -1):
            ct = self.c[d]
            if ct == 0:
                continue
            abs_ct = abs(ct)
            if d == 0:
                t = str(abs_ct)
            elif d == 1:
                if abs_ct == 1:
                    t = "x"
                else:
                    t = f"{abs_ct}x"
            else: 
                if abs_ct == 1:
                    t = f"x^{d}"
                else:
                    t = f"{abs_ct}x^{d}"
            if len(p) == 0:
                if ct < 0:
                    p.append("-" + t) 
                else:
                    p.append(t)
            else:
                if ct < 0:
                    p.append(" - " + t)
                else:
                    p.append(" + " + t)
        if len(p) == 0:
            return "0"
        return "".join(p)

    def __eq__(self,o):
        if isinstance(o, (int, float)):
            o = Polynomial(o)
        if not isinstance(o, Polynomial):
            return NotImplemented
        return self.c == o.c

    def __add__(self, o):
        if isinstance(o, (int, float)):
            o = Polynomial(o)

        if not isinstance(o, Polynomial):
            return NotImplemented

        result = []

        for d in range(max(len(self.c), len(o.c))):
            a = self.c[d] if d < len(self.c) else 0
            b = o.c[d] if d < len(o.c) else 0
            result.append(a + b)

        return Polynomial(result)
    def __radd__(self,o):
        return self + o

    def __neg__(self):
        return Polynomial([-ct for ct in self.c])
    
    def __sub__(self, o):
        if isinstance(o, (int, float)):
            o = Polynomial(o)

        if not isinstance(o, Polynomial):
            return NotImplemented

        return self + (-o)

    def __rsub__(self, o):
        if isinstance(o, (int, float)):
            o = Polynomial(o)

        if not isinstance(o, Polynomial):
            return NotImplemented

        return o + (-self)
    
    def __call__(self,x):
        r = 0
        for i, ct in enumerate(self.c):
            r += ct * x ** i
        return r
    
    def degree(self):
        return len(self.c) - 1

    def der(self, d = 1):
        c = self.c.copy()
        if d < 0:
            return False 
        o = 0
        while d > o:

            if len(c) == 1:
                return Polynomial(0)
            
            nc = []

            for i in range(1,len(c)):
                nc.append(c[i] * i)
            c = nc 
            o += 1
        return Polynomial(c)

    def __mul__(self,o):
        if isinstance(o, (int, float)):
            return Polynomial([t * o for t in self.c])
        if not isinstance(o, Polynomial):
            return NotImplemented
        result = [0] * (len(self.c) + len(o.c) - 1)
        for i, a in enumerate(self.c):
            for j, b in enumerate(o.c):
                result[i + j] += a * b
        return Polynomial(result)
    
    def __rmul__(self,o):
        return self * o
    
    def __iter__(self):
        self.i = 0
        return self
    
    def __next__(self):
        if self.i < len(self.c):
            result = (self.i,self.c[self.i])
            self.i += 1
            return result 
        else:
            raise StopIteration
````

### Разбор кода:

````python
class Polynomial: # Класс для работы с многочленами

    def __init__(self, *c): # Конструктор, *c собирает все аргументы в кортеж
        if len(c) == 1 and isinstance(c[0], list): # Проверяем, передан ли один список
            self.c = c[0].copy() # Копируем список коэффициентов
        elif len(c) == 1 and isinstance(c[0], dict): # Проверяем, передан ли словарь
            data = c[0] # Сохраняем словарь в переменную data
            if not data: # Проверяем, пустой ли словарь
                self.c = [0] # Если пустой, создаём нулевой многочлен
            else: # Если словарь содержит элементы
                maxd = max(data.keys()) # Находим максимальную степень среди ключей
                self.c = [0] * (maxd + 1) # Создаём список нулей нужной длины
            for d, ct in data.items(): # Перебираем пары степень-коэффициент
                self.c[d] = ct # Записываем коэффициент по индексу его степени
        elif len(c) == 1 and isinstance(c[0], Polynomial): # Проверяем, передан ли многочлен
            self.c = c[0].c.copy() # Копируем коэффициенты другого многочлена
        else: # Если аргументы переданы другим способом
            self.c = list(c) # Преобразуем аргументы в список
        if len(self.c) == 0: # Проверяем, пустой ли список коэффициентов
            self.c = [0] # Заменяем пустой список на [0]
        while len(self.c) > 1 and self.c[-1] == 0: # Пока последний коэффициент ноль и элементов больше одного
            self.c.pop() # Удаляем последний ноль, чтобы убрать лишние старшие степени

    def __repr__(self): # Метод технического представления объекта
        return f"Polynomial {self.c}" # Возвращаем строку вида Polynomial [1, 2, 3]

    def __str__(self): # Метод красивого вывода многочлена
        p = [] # Создаём список для хранения слагаемых в виде строк
        for d in range(len(self.c) - 1, -1, -1): # Перебираем степени от старшей к младшей
            ct = self.c[d] # Получаем коэффициент при степени d
            if ct == 0: # Проверяем, равен ли коэффициент нулю
                continue # Пропускаем нулевое слагаемое
            abs_ct = abs(ct) # Получаем модуль коэффициента без знака
            if d == 0: # Если степень нулевая, это свободный член
                t = str(abs_ct) # Преобразуем коэффициент в строку
            elif d == 1: # Если степень равна единице
                if abs_ct == 1: # Если модуль коэффициента равен единице
                    t = "x" # Записываем только x, без коэффициента 1
                else: # Если модуль коэффициента не равен единице
                    t = f"{abs_ct}x" # Записываем коэффициент и x
            else: # Если степень больше единицы
                if abs_ct == 1: # Если модуль коэффициента равен единице
                    t = f"x^{d}" # Записываем x в степени d без коэффициента 1
                else: # Если модуль коэффициента не равен единице
                    t = f"{abs_ct}x^{d}" # Записываем коэффициент и степень x
            if len(p) == 0: # Проверяем, первое ли это ненулевое слагаемое
                if ct < 0: # Если коэффициент отрицательный
                    p.append("-" + t) # Добавляем минус перед первым слагаемым
                else: # Если коэффициент положительный
                    p.append(t) # Добавляем первое слагаемое без плюса
            else: # Если в списке уже есть слагаемые
                if ct < 0: # Если коэффициент отрицательный
                    p.append(" - " + t) # Добавляем слагаемое со знаком минус
                else: # Если коэффициент положительный
                    p.append(" + " + t) # Добавляем слагаемое со знаком плюс
        if len(p) == 0: # Проверяем, остался ли список пустым
            return "0" # Если все коэффициенты нулевые, возвращаем строку "0"
        return "".join(p) # Объединяем все слагаемые в одну строку

    def __eq__(self, o): # Метод сравнения объектов через оператор ==
        if isinstance(o, (int, float)): # Проверяем, является ли второй аргумент числом
            o = Polynomial(o) # Преобразуем число в постоянный многочлен
        if not isinstance(o, Polynomial): # Проверяем, что аргумент является многочленом
            return NotImplemented # Сообщаем Python, что этот тип не поддерживается
        return self.c == o.c # Сравниваем списки коэффициентов и возвращаем True или False

    def __add__(self, o): # Метод сложения многочленов через оператор +
        if isinstance(o, (int, float)): # Проверяем, является ли второй аргумент числом
            o = Polynomial(o) # Преобразуем число в многочлен
        if not isinstance(o, Polynomial): # Проверяем тип второго аргумента
            return NotImplemented # Сообщаем, что сложение с этим типом не реализовано
        result = [] # Создаём пустой список для коэффициентов суммы
        for d in range(max(len(self.c), len(o.c))): # Перебираем степени до длины большего многочлена
            a = self.c[d] if d < len(self.c) else 0 # Берём коэффициент первого многочлена или 0
            b = o.c[d] if d < len(o.c) else 0 # Берём коэффициент второго многочлена или 0
            result.append(a + b) # Складываем коэффициенты одинаковых степеней
        return Polynomial(result) # Возвращаем новый многочлен с полученными коэффициентами

    def __radd__(self, o): # Метод обратного сложения, например 5 + p
        return self + o # Вызываем обычное сложение, поменяв порядок операндов

    def __neg__(self): # Метод унарного минуса, например -p
        return Polynomial([-ct for ct in self.c]) # Меняем знак каждого коэффициента и создаём новый многочлен

    def __sub__(self, o): # Метод вычитания через оператор -
        if isinstance(o, (int, float)): # Проверяем, является ли второй аргумент числом
            o = Polynomial(o) # Преобразуем число в многочлен
        if not isinstance(o, Polynomial): # Проверяем, является ли аргумент многочленом
            return NotImplemented # Сообщаем, что вычитание с этим типом не поддерживается
        return self + (-o) # Вычитаем через сложение с противоположным многочленом

    def __rsub__(self, o): # Метод обратного вычитания, например 5 - p
        if isinstance(o, (int, float)): # Проверяем, является ли левый аргумент числом
            o = Polynomial(o) # Преобразуем число в многочлен
        if not isinstance(o, Polynomial): # Проверяем тип левого аргумента
            return NotImplemented # Сообщаем, что операция с этим типом не поддерживается
        return o + (-self) # Из левого аргумента вычитаем текущий многочлен

    def __call__(self, x): # Позволяет вызывать объект как функцию, например p(2)
        r = 0 # Создаём переменную для суммы значений слагаемых
        for i, ct in enumerate(self.c): # Перебираем индексы-степени и коэффициенты
            r += ct * x ** i # Вычисляем коэффициент * x в степени i и прибавляем к сумме
        return r # Возвращаем значение многочлена в точке x

    def degree(self): # Метод определения степени многочлена
        return len(self.c) - 1 # Возвращаем индекс последнего коэффициента, для нулевого многочлена это 0

    def der(self, d=1): # Метод вычисления производной порядка d, по умолчанию первой
        c = self.c.copy() # Копируем коэффициенты, чтобы не изменять исходный многочлен
        if d < 0: # Проверяем, отрицательный ли порядок производной
            return False # Если порядок отрицательный, возвращаем False
        o = 0 # Счётчик количества вычисленных производных
        while d > o: # Повторяем, пока не вычислим производную нужного порядка
            if len(c) == 1: # Если многочлен стал постоянным
                return Polynomial(0) # Возвращаем ноль, так как производная константы равна нулю
            nc = [] # Создаём список для коэффициентов новой производной
            for i in range(1, len(c)): # Перебираем степени начиная с первой, пропуская свободный член
                nc.append(c[i] * i) # Умножаем коэффициент на степень согласно правилу производной
            c = nc # Заменяем коэффициенты на коэффициенты полученной производной
            o += 1 # Увеличиваем счётчик вычисленных производных
        return Polynomial(c) # Возвращаем многочлен с производной нужного порядка

    def __mul__(self, o): # Метод умножения многочленов через оператор *
        if isinstance(o, (int, float)): # Проверяем, является ли второй аргумент числом
            return Polynomial([t * o for t in self.c]) # Умножаем каждый коэффициент на число
        if not isinstance(o, Polynomial): # Проверяем, является ли второй аргумент многочленом
            return NotImplemented # Сообщаем, что умножение с этим типом не поддерживается
        result = [0] * (len(self.c) + len(o.c) - 1) # Создаём список нулей для коэффициентов произведения
        for i, a in enumerate(self.c): # Перебираем степени и коэффициенты первого многочлена
            for j, b in enumerate(o.c): # Перебираем степени и коэффициенты второго многочлена
                result[i + j] += a * b # Умножаем коэффициенты и складываем степени, суммируя подобные члены
        return Polynomial(result) # Возвращаем новый многочлен с результатом умножения

    def __rmul__(self, o): # Метод обратного умножения, например 5 * p
        return self * o # Вызываем обычное умножение

    def __iter__(self): # Метод запуска итерации, вызывается циклом for
        self.i = 0 # Устанавливаем индекс текущего коэффициента равным нулю
        return self # Возвращаем сам объект как итератор

    def __next__(self): # Метод получения следующего элемента итерации
        if self.i < len(self.c): # Проверяем, остались ли коэффициенты для перебора
            result = (self.i, self.c[self.i]) # Создаём кортеж (степень, коэффициент)
            self.i += 1 # Увеличиваем индекс для следующего вызова
            return result # Возвращаем текущую пару
        else: # Если все коэффициенты уже перебраны
            raise StopIteration # Сообщаем Python о завершении итерации
````
